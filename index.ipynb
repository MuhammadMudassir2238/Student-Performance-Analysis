```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
# Load Dataset
data = pd.read_csv("dataset.csv")
```

```python
# display(data.head())
# display(data.shape)
# display(data.info())
# display(data.describe())
display(data)
```

```python
# average 
data["average"] = data[["math","english","physics"]].mean(axis=1)
data["Total_Marks"] = data[["math","english","physics"]].sum(axis=1)
display(data)
```

```python
# Top Students
top_student = data.loc[data["average"].idxmax()]
display(top_student)
```

```python
plt.bar(data["student"], data["average"])
plt.xlabel("Student")
plt.ylabel("Average Marks")
plt.title("Student performance")
plt.show()
```
